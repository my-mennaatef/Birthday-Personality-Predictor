```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv(r"C:\Users\Asus\OneDrive\Desktop\birthday project\birthday_data.csv")
```

```python
df.head()
```

```python
#label encoding
# convert to numbers for the model to understand
#0-->no
#1-->yes

le = LabelEncoder()

df["likes_party"] = le.fit_transform(df["likes_party"])
df["favorite_activity"] = le.fit_transform(df["favorite_activity"])
df["favorite_color"] = le.fit_transform(df["favorite_color"])
df["personality"] = le.fit_transform(df["personality"])
```

```python
df.head()
```

```python
# feature and target

X = df.drop("personality", axis=1)
y = df["personality"]
```

```python
# train model

model = DecisionTreeClassifier()
model.fit(X, y)
```

```python
# predication

prediction = model.predict([[20,1,3,2]])
print(prediction)
```

```python
# more simple way

new_data = pd.DataFrame(
    [[20,1,3,2]],
    columns=[
        "age",
        "likes_party",
        "favorite_activity",
        "favorite_color"
    ]
)
```

```python
#test

prediction = model.predict(new_data)
print(prediction)
```

```python
personality_names = {
    0: "calm",
    1: "creative",
    2: "energetic",
    3: "introvert",
    4: "social"
}
```

```python
# test

print(personality_names[prediction[0]])
```
